```python
import pandas as pd
import numpy as np
```

```python
full_matchups = pd.read_csv("Tournament Matchups.csv")
display(full_matchups)
```

```python
matchups = full_matchups.copy()
matchups['temp_col'] = 1
matchups['match_num'] = (matchups.groupby('YEAR')['temp_col'].cumsum() - 1) // 2 + 1
display(matchups)
```

```python
recent = matchups[matchups['YEAR'] == 2025]
display(recent)
```

```python
results_dict = {}

matchups['seed_pairing'] = matchups.groupby(['match_num', 'YEAR'])['SEED'].transform('sum') - matchups['SEED']
winner_seed = matchups.loc[matchups.groupby(['match_num', 'YEAR'])['SCORE'].idxmax(), ['match_num', 'YEAR', 'SEED']]
winner_seed = winner_seed.set_index(['match_num', 'YEAR'])['SEED']

matchups['winner_seed'] = matchups.set_index(['match_num', 'YEAR']).index.map(winner_seed)
matchups = matchups.drop_duplicates(subset=['match_num', 'YEAR'])
display(matchups)

for _, row in matchups.iterrows():
    if f"{row['SEED']}-{row['seed_pairing']}" not in results_dict:
        results_dict[f"{row['SEED']}-{row['seed_pairing']}"] = [0, 0]

    if row['winner_seed'] == row['SEED']:
        results_dict[f"{row['SEED']}-{row['seed_pairing']}"][0] += 1
    else:
        results_dict[f"{row['SEED']}-{row['seed_pairing']}"][1] += 1

print(results_dict)
```

```python
new_matchups = pd.DataFrame(columns=['seed', 'curr_round', 'match_num', 'seed_pairing'])
new_matchups['seed'] = [1, 16, 8, 9, 5, 12, 4, 13, 6, 11, 3, 14, 7, 10, 2, 15]*4
new_matchups['curr_round'] = 64
new_matchups['temp_col'] = 1
new_matchups['match_num'] = (new_matchups['temp_col'].cumsum() - 1) // 2 + 1
new_matchups['seed_pairing'] = new_matchups.groupby(['match_num', 'curr_round'])['seed'].transform('sum') - new_matchups['seed']
new_matchups = new_matchups.drop_duplicates(subset=['match_num', 'curr_round'])
random_beta = lambda x: np.random.beta(results_dict.get(f"{x['seed']}-{x['seed_pairing']}", [1, 1]), results_dict.get(f"{x['seed']}-{x['seed_pairing']}", [1, 1])[1] + 1)
random_winner = lambda x: np.random.choice([x['seed'], x['seed_pairing']], p=[random_beta(x)[0], 1-random_beta(x)[0]])
new_matchups['winner'] = new_matchups.apply(random_winner, axis=1)
new_matchups.drop(columns=['temp_col'], inplace=True)
display(new_matchups)
```
